# practice atleast 2 example for each using ndim, shape, size and itemsize

```python
import numpy as np

arr=np.array([1,2,3])

arr2=np.array([[10,20,30],
              [60,80,90]])

arr3=np.array([
    [11,21,31],
    [2,3,42],
    [13,23,33]
])

arr4=np.array([
    [20,30,40],
    [50,60,70],
    [1,2,3],
    [4,5,6]
])

print("arr")
print(arr)

print("ndim:", arr.ndim)
print("shape:", arr.shape)
print("size:", arr.size)
print("itemsize:", arr.itemsize)

print("\narr2")
print(arr2)

print("ndim:", arr2.ndim)
print("shape:", arr2.shape)
print("size:", arr2.size)
print("itemsize:", arr2.itemsize)

print("\narr3")
print(arr3)

print("ndim:", arr3.ndim)
print("shape:", arr3.shape)
print("size:", arr3.size)
print("itemsize:", arr3.itemsize)

print("\narr4")
print(arr4)

print("ndim:", arr4.ndim)
print("shape:", arr4.shape)
print("size:", arr4.size)
print("itemsize:", arr4.itemsize)
```

# find average total_bill and average tip

```python
import pandas as pd

tips = pd.read_csv("/kaggle/input/datasets/saurabhbadole/restaurant-tips-dataset/tips.csv")

print("average total_bill:", tips["total_bill"].mean())
print("average tip:", tips["tip"].mean())
```

# find standard deviasion and variance of tip

```python
print("Standard deviation of tip:", tips["tip"].std())
print("Variance of tip:", tips["tip"].var())
```

# find log values of tip

```python
import numpy as np
import pandas as pd

tips = pd.read_csv(
    "/kaggle/input/datasets/saurabhbadole/restaurant-tips-dataset/tips.csv"
)

tip = tips["tip"].to_numpy()

print("Tip array:")
print(tip)

print("\nLog values:")
print(np.log(tip))
```
